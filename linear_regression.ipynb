# Lets make the linear regression class

```python
import numpy as np

class LinearRegressionGD:
    def __init__(self, lr=0.01, n_epochs=1000):
        self.lr = lr
        self.n_epochs = n_epochs
        self.W = None
        self.b = None

    def predict(self, X):
        return X @ self.W + self.b

    def loss(self, y_pred, y_true):
        n = y_true.shape[0]
        return np.sum((y_pred - y_true) ** 2) / n

    def gradient(self, X, y_true):
        n = X.shape[0]
        y_pred = self.predict(X)
        error = y_pred - y_true

        dW = (2 / n) * (X.T @ error)
        db = (2 / n) * np.sum(error)

        return dW, db

    def fit(self, X, y):
        n_samples, n_features = X.shape
        self.W = np.zeros((n_features, 1))
        self.b = 0.0

        for epoch in range(self.n_epochs):
            dW, db = self.gradient(X, y)
            self.W -= self.lr * dW
            self.b -= self.lr * db

            if (epoch + 1) % (self.n_epochs // 10) == 0:
                y_pred = self.predict(X)
                current_loss = self.loss(y_pred, y)
                print(f"Epoch {epoch+1}/{self.n_epochs} - Loss: {current_loss:.4f}")
```

# Lets generate a dummy data and train the lin_regression model

```python
# create data
np.random.seed(0)
X = np.linspace(0, 10, 50).reshape(-1, 1)
y = 3 * X + 2 + np.random.randn(50, 1)

# create and train model
model = LinearRegressionGD(lr=0.01, n_epochs=1000)
model.fit(X, y)

print("Learned W:", model.W)
print("Learned b:", model.b)

# make predictions
x_test = np.array([[0], [2], [5]])
y_pred = model.predict(x_test)
print("Predictions:\n", y_pred)
```
